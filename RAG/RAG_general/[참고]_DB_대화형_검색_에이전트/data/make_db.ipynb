```python
import sqlite3
import pandas as pd

conn = sqlite3.connect("titanic.db")
cur = conn.cursor()

# 2) 테이블 생성
cur.execute("""
CREATE TABLE IF NOT EXISTS passengers (
    PassengerId INTEGER PRIMARY KEY,
    Survived INTEGER,
    Pclass INTEGER,
    Name TEXT,
    Sex TEXT,
    Age REAL,
    SibSp INTEGER,
    Parch INTEGER,
    Ticket TEXT,
    Fare REAL,
    Cabin TEXT,
    Embarked TEXT
)
""")

df = pd.read_csv("titanic.csv")

df.to_sql("passengers", conn, if_exists="replace", index=False)

cur = conn.cursor()


cur.execute("SELECT Name, Survived, Pclass FROM passengers LIMIT 5;")
rows = cur.fetchall()
for row in rows:
    print(row)

conn.close()
```
